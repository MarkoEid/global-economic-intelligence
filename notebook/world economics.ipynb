# Scrab data from Wikipedia

```python
import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/List_of_countries_by_real_GDP_growth_rate"

headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)
tables = pd.read_html(response.text)

print(len(tables))   
GDP_Gr = tables[1]    
GDP_Gr.columns = ["name", "GDP Growth"]
GDP_Gr.head()
```

```python
GDP_Gr2=df[["name","GDP Growth"]]
GDP_Gr2
```

```python
GDP_Gr3=GDP_Gr.iloc[2:]
GDP_Gr3
```

```python
GDP_Gr3 = GDP_Gr3[GDP_Gr3["GDP Growth"] != "Rate (%)"]
```

```python
GDP_Gr3["GDP Growth"] = GDP_Gr3["GDP Growth"].str.replace("−", "-", regex=True)
GDP_Gr3["GDP Growth"]=GDP_Gr3["GDP Growth"].astype(float)
```
